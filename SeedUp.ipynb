# **🚀 Step 1: Project Setup**

Clone the repository, install all dependencies, and verify the environment — all in one step.

```python
#@title **Project Setup — Clone, Install & Verify** { display-mode: "form" }
#@markdown Click ▶️ to clone the repo, install dependencies, and verify the environment.

#@markdown ### ⚙️ Repository Configuration
REPO_URL = "https://github.com/KulanS/SeedUp.git" #@param {type:"string"}
BRANCH = "develop" #@param {type:"string"}

# Derive repo name from URL
REPO_NAME = REPO_URL.rstrip('/').split('/')[-1].replace('.git', '')
import os
import sys
import time
import subprocess

# ─── Helpers ────────────────────────────────────────────────────────

_start_time = time.time()
_phase_results = {}  # track pass/fail per phase

def _banner(title, icon="📌"):
    width = 60
    print(f"\n{'━' * width}")
    print(f"{icon}  {title}")
    print(f"{'━' * width}")

def _log(msg, indent=1):
    prefix = "  " * indent
    print(f"{prefix}▸ {msg}")

def _ok(msg):
    print(f"  ✅ {msg}")

def _fail(msg):
    print(f"  ❌ {msg}")

def _elapsed():
    return f"{time.time() - _start_time:.1f}s"

# ─── Phase 1: Clone Repository ──────────────────────────────────────

_banner("Phase 1/3 — Clone Repository", "📥")

try:
    # Clean up previous clone if it exists
    if os.path.exists(REPO_NAME):
        _log(f"Removing existing {REPO_NAME} directory…")
        !rm -rf {REPO_NAME}
        _log("Cleaned up previous clone.")

    _log(f"Cloning {REPO_URL} …")
    !git clone {REPO_URL} 2>&1

    if not os.path.isdir(REPO_NAME):
        raise RuntimeError(f"{REPO_NAME} directory not found after git clone.")

    %cd {REPO_NAME}

    _log(f"Switching to {BRANCH} branch…")
    !git switch {BRANCH} 2>&1

    _log(f"Working directory: {os.getcwd()}")
    _ok("Repository cloned successfully.")
    _phase_results['Clone'] = True

except Exception as e:
    _fail(f"Clone failed: {e}")
    _phase_results['Clone'] = False

# ─── Phase 2: Install Dependencies ──────────────────────────────────

_banner("Phase 2/3 — Install Dependencies", "📦")

try:
    req_file = 'requirements.txt'
    if not os.path.isfile(req_file):
        raise FileNotFoundError(f"{req_file} not found in {os.getcwd()}")

    _log(f"Found {req_file}")

    # Show what will be installed
    with open(req_file) as f:
        deps = [l.strip() for l in f if l.strip() and not l.startswith('#')]
    _log(f"Packages to install: {', '.join(deps)}")

    _log("Running pip install (this may take ~2 minutes)…")
    result = subprocess.run(
        [sys.executable, '-m', 'pip', 'install', '-r', req_file],
        capture_output=True, text=True
    )

    if result.returncode != 0:
        _fail("pip install returned a non-zero exit code.")
        print(result.stderr[-1500:] if len(result.stderr) > 1500 else result.stderr)
        _phase_results['Install'] = False
    else:
        # Count installed vs already-satisfied
        installed = result.stdout.count('Successfully installed')
        satisfied = result.stdout.count('already satisfied')
        _log(f"pip output: {installed} new install(s), {satisfied} already satisfied.")
        _ok("All dependencies installed.")
        _phase_results['Install'] = True

except Exception as e:
    _fail(f"Install failed: {e}")
    _phase_results['Install'] = False

# ─── Phase 3: Verify Installations ──────────────────────────────────

_banner("Phase 3/3 — Verify Installations", "🔍")

try:
    # Python version
    py_version = sys.version.split()[0]
    _log(f"Python version: {py_version}")

    # libtorrent
    failing_message = "libtorrent is NOT installed."
    import libtorrent as lt
    _log(f"libtorrent version: {lt.version}")
 
    # Google API clients
    failing_message = "Google API clients are NOT installed."
    from googleapiclient import discovery
    from google.oauth2.credentials import Credentials
    _log("Google API clients: available")

    # Project entry-point
    if os.path.isfile('main.py'):
        _log("main.py: found")
    else:
        _fail("main.py not found in project root.")
        raise FileNotFoundError("main.py")

    _ok("All verifications passed.")
    _phase_results['Verify'] = True

except ImportError:
    _fail(failing_message)
    raise
except Exception as e:
    _fail(f"Verification failed: {e}")
    _phase_results['Verify'] = False

# ─── Summary ────────────────────────────────────────────────────────

_banner("Setup Summary", "📋")

all_passed = True
for phase, passed in _phase_results.items():
    status = "✅ PASS" if passed else "❌ FAIL"
    print(f"  {phase:.<30s} {status}")
    if not passed:
        all_passed = False

print(f"\n  ⏱  Total elapsed: {_elapsed()}")

if all_passed:
    print("\n🎉 Environment is ready — proceed to the next step!")
else:
    print("\n⚠️  One or more phases failed. Review the logs above before continuing.")

    
```

# **🔍 Step 3: Preview Torrent Contents**

Preview the torrent file structure, sizes, and Google Drive upload status before downloading.

- **Check Drive Status:** Cross-reference with Google Drive to see what's already uploaded
- **File Index:** Use the index numbers to selectively download in the next step

```python
#@title **Preview Torrent Contents** { display-mode: "form" }

#@markdown ### 🔗 Enter Torrent Information
#@markdown Paste magnet link OR path to .torrent file:
TORRENT_SOURCE = "" #@param {type:"string"}

#@markdown ### ⚙️ Options
CHECK_DRIVE_STATUS = True #@param {type:"boolean"}

if not TORRENT_SOURCE:
    print("⚠️ Please enter a magnet link or torrent file path!")
else:
    cmd = f'python main.py preview -t "{TORRENT_SOURCE}"'
    if CHECK_DRIVE_STATUS:
        cmd += " --check-drive"
    !{cmd}
```

# **🌐 Step 4: Download Torrent**

Download selected files from the torrent and automatically upload to Google Drive.

### Options:
- **Selected Files:** Enter `all` or comma-separated indices from preview (e.g. `0,2,5`)
- **Skip Existing:** Skip files that already exist in Drive

**Note:** Files are uploaded incrementally as they complete and deleted locally to save disk space.

```python
#@title **Download Torrent** { display-mode: "form" }

#@markdown ### 🔗 Enter Torrent Information
#@markdown Paste magnet link OR path to .torrent file:
TORRENT_SOURCE = "" #@param {type:"string"}

#@markdown ### ⚙️ Options
SELECTED_FILES = "all" #@param {type:"string"}
#@markdown Enter "all" or comma-separated indices from preview (e.g. "0,2,5")
SKIP_EXISTING = True #@param {type:"boolean"}

if not TORRENT_SOURCE:
    print("⚠️ Please enter a magnet link or torrent file path!")
else:
    print("🚀 Starting incremental download & upload pipeline...\n")
    print("📁 Files will be uploaded to 'SeedUp Downloads' folder in Drive root\n")

    # Build command
    cmd = f'python main.py download -t "{TORRENT_SOURCE}" --upload --select {SELECTED_FILES}'

    if not SKIP_EXISTING:
        cmd += " --no-skip"

    # Execute pipeline
    !{cmd}
```

# **📤 Step 5: Upload Existing Files**

If you already have downloaded files and want to upload them to Google Drive separately.

**Note:** Files will be uploaded to the 'SeedUp Downloads' folder automatically (or your custom folder if specified)

```python
#@title **Upload Files to Drive** { display-mode: "form" }

#@markdown ### 📁 File/Folder to Upload
LOCAL_PATH = "" #@param {type:"string"}

#@markdown ### ⚙️ Options
SKIP_EXISTING_FILES = True #@param {type:"boolean"}

import os

if not LOCAL_PATH:
    print("⚠️ Please enter a file or folder path to upload!")
elif not os.path.exists(LOCAL_PATH):
    print(f"⚠️ Path does not exist: {LOCAL_PATH}")
else:
    print("📤 Starting upload to Google Drive...\n")
    print("📁 Files will be uploaded to 'SeedUp Downloads' folder in Drive root\n")

    # Build command
    cmd = f"python main.py upload -p \"{LOCAL_PATH}\""

    if not SKIP_EXISTING_FILES:
        cmd += " --no-skip"

    # Execute upload
    !{cmd}
```

## 📚 **Additional Information**

### Troubleshooting

**Download is slow:**
- This depends on the number of seeders and your torrent health
- Colab's network speed varies
- Try using different trackers or magnet links

**Upload failing:**
- Check your Google Drive has enough space
- Re-authenticate if needed
- Check if files already exist when using skip option
- The SeedUp folder is created automatically - no manual setup needed

**Session disconnected:**
- Downloads are saved and can be resumed
- Use the status check to verify saved sessions
- Run the download command again with the same torrent
- Clear stuck sessions if needed

### Best Practices

1. **For large downloads:**
   - Monitor the session and be ready to resume if it disconnects
   - Use the keep-alive cell to prevent idle disconnects
   - Split very large downloads into smaller parts

2. **Storage management:**
   - All files are organized in the 'SeedUp Downloads' folder automatically
   - Clear the downloads folder after successful upload
   - Monitor Colab's disk space usage
   - Use `skip-existing` to avoid duplicate uploads

3. **Finding your uploads:**
   - Open Google Drive and look for the "SeedUp Downloads" folder
   - All uploaded files will be there
   - The folder link is displayed after each upload

### Session Management

The project uses a sophisticated session management system that:
- Saves download progress automatically
- Enables resuming from the exact point of interruption
- Maintains tracker and peer information
- Cleans up automatically after successful completion
